# Problem Statement

Agricultural activities are influenced by seasonal variations in environmental conditions, farming practices, resource availability and market conditions. As a result, agricultural performance may differ from one season to another.

However, raw agricultural data does not clearly explain how agricultural performance changes across seasons or what patterns can be observed in different seasonal conditions.

The problem is to analyze the given agricultural dataset and investigate seasonal differences in agricultural performance by identifying meaningful patterns, trends, relationships and variations within the available data.

```python
# Data Quality Assessment and Cleaning
# Missing values
print("Missing Values:")
print(df.isnull().sum())
```

```python
# Duplicate records
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Create a copy of the dataset
df_clean = df.copy()

# Fill missing values in numerical columns with median
numeric_columns = df_clean.select_dtypes(include=np.number).columns

for column in numeric_columns:
    if df_clean[column].isnull().sum() > 0:
        df_clean[column] = df_clean[column].fillna(
            df_clean[column].median()
        )

print("Missing values after cleaning:")
print(df_clean.isnull().sum().sum())
```

```python
# Remove duplicate rows
df_clean = df_clean.drop_duplicates()

print("Duplicates after cleaning:", df_clean.duplicated().sum())

print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)
```

# Key Insights

1. **Seasonal Yield:** The Kharif season recorded the highest average agricultural yield at 5.63 tonnes/hectare, followed by Rabi at 5.04 tonnes/hectare, while Zaid recorded the lowest average yield at 4.64 tonnes/hectare.

2. **Seasonal Profit:** The Kharif season was the most profitable, generating an average profit of ₹178,914.65 per farm. In contrast, Zaid recorded a negative average profit of ₹24,804.82 per farm, indicating that average costs exceeded average revenue during the observed period.

3. **Seasonal Production:** Kharif recorded the highest average production of 46.31 tonnes per farm, followed by Rabi with 41.49 tonnes per farm and Zaid with 38.89 tonnes per farm. This indicates that agricultural production performance was highest during the Kharif season in the given dataset.

4. **Rainfall and Yield:** A very weak positive linear relationship was observed between rainfall and crop yield (r = 0.031), indicating that rainfall alone had little linear association with yield in the dataset.

5. **Temperature and Yield:** Average temperature showed virtually no linear correlation with crop yield (r = 0.009), suggesting that temperature variation within the observed dataset had little linear association with per-hectare yield.

6. **Crop Performance:** Sugarcane demonstrated the highest yield performance among the crops analyzed, reaching a peak average yield of 53.46 tonnes/hectare during Kharif. Pulses recorded the lowest average yield, ranging from 0.65 to 1.04 tonnes/hectare across seasons.

7. **Irrigation Method:** Drip irrigation achieved the highest average yield at 6.62 tonnes/hectare, followed by Sprinkler at 5.19 tonnes/hectare, Flood irrigation at 4.90 tonnes/hectare, and Rainfed farming at 4.61 tonnes/hectare.

8. **Water Efficiency and Profit:** Water efficiency showed a moderate positive correlation with farm profitability (r = 0.489), indicating that farms with higher water efficiency tended to have higher profitability.

9. **Disease/Pest Risk and Yield:** Disease and pest risk showed almost no linear correlation with crop yield (r = 0.013), suggesting that pest-risk variation had little observable linear association with yield in the analyzed dataset.

10. **Correlation Analysis:** The strongest correlation observed in the dataset was between crop yield and water efficiency (r = 0.915), indicating a very strong positive association. Revenue also showed a strong positive correlation with net profit (r = 0.887).

# Evidence-Based Recommendations

Based on the analysis and key insights obtained from the dataset, the following recommendations can be made:

1. **Seasonal Crop Planning:** Agricultural planning should consider seasonal variations in environmental conditions such as rainfall and temperature when selecting suitable crops and planning farming activities.

2. **Season-Specific Crop Selection:** Crops that demonstrate higher yields during specific seasons can be prioritized during those periods to improve agricultural productivity.

3. **Efficient Irrigation Methods:** Irrigation methods associated with higher average yields, such as drip irrigation, can be considered where water availability and farming conditions permit.

4. **Improve Water Efficiency:** Water-efficient farming practices should be encouraged to improve productivity while reducing unnecessary water consumption and supporting better resource management.

5. **Focus on Profitable Seasons:** Seasons showing higher average profitability, particularly Kharif in this dataset, can be prioritized for suitable crops and farming activities.

6. **Disease and Pest Monitoring:** Disease and pest risks should be monitored regularly, especially during periods where higher risk is observed alongside lower crop yields.

7. **Seasonal Resource Allocation:** Resources such as water, irrigation, and farm inputs should be allocated according to seasonal requirements and observed agricultural performance.

8. **Use Broader Data for Future Planning:** Further analysis using larger datasets along with external weather, soil, and market information can provide more reliable insights and support improved agricultural planning.

# Limitations

Although the analysis provides useful insights into seasonal agricultural performance, the project has certain limitations:

1. **Dataset Limitations:** The analysis is limited to the variables available in the provided dataset and may not cover all factors that influence agricultural performance.

2. **Limited Generalization:** The findings are based on historical records and may not represent all agricultural regions, farming practices, or environmental conditions.

3. **External Factors:** Factors such as government policies, sudden climate events, changes in input costs, and market fluctuations may not be fully captured in the dataset.

4. **No Causal Conclusions:** The analysis identifies relationships and patterns between variables but does not establish direct cause-and-effect relationships.

5. **Missing Value Imputation:** Missing numerical values were handled using median imputation, which may affect the interpretation of variables where missing data was present.

6. **Limited External Data:** The analysis does not incorporate additional real-world weather, soil, or market data, which could provide deeper insights into agricultural performance.
